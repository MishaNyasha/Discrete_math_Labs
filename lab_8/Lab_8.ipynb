```python
def analyze(w):
    seen_c = seen_d = False
    for ch in w:
        if ch == 'a':
            return False, "Содержит 'a'"
        elif ch == 'c':
            seen_c = True
        elif ch == 'd':
            seen_d = True
        elif ch == 'b':
            pass
        else:
            return False, f"Недопустимый символ '{ch}'"
    if not seen_c and not seen_d:
        return False, "нету 'c' и 'd'"
    if not seen_c:
        return False, "нету 'c'"
    if not seen_d:
        return False, "нету 'd'"
    return True, "Содержит 'c' и 'd' и нету 'a'"

if __name__ == "__main__":
    words = ["abcd", "bcd", "cd", "bc", "cb", "a"]
    for w in words:
        ok, reason = analyze(w)
        print(f"{w}: {'YES' if ok else 'NO'} ({reason})")
```
